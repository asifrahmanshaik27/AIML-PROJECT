```python
import numpy as np
marks = np.random.randint(0, 100,500)
between_60_80 = 0
above_90 = 0
below_50 = 0

for mark in marks:
    if 60 <= mark <= 80:
        between_60_80 += 1
    if mark > 90:
        above_90 += 1
    if mark < 50:
        below_50 += 1

print("Total Students:", len(marks))
print("Students scoring between 60 and 80:", between_60_80)
print("Students scoring above 90:", above_90)
print("Students scoring below 50:", below_50)
```

Outliers
27.07.2026


```python
df.columns
```

```python
q1=df['Experience_Years'].quantile(0.25)
q3=df["Experience_Years"].quantile(0.75)
IQR=q3-q1
lower_limit=q1-1.5*IQR
upper_limit=q3+1.5*IQR
print("q1:",q1)
print('q3:',q3)
print("IQR:",IQR)
print("lower_limit:",lower_limit)
print("upper_limit:",upper_limit)
outliers=df[
    (df["Experience_Years"]<lower_limit) |
    (df["Experience_Years"]>upper_limit)
]
print("\n EXperience Outliers:",len(outliers))
print("\nExperience outliers:")
print(outliers[["Worker_ID","Experience_Years"]].head(20))
```

```python
q1=df['Salary'].quantile(0.25)
q3=df["Salary"].quantile(0.75)
IQR=q3-q1
lower_limit=q1-1.5*IQR
upper_limit=q3+1.5*IQR
print("q1:",q1)
print('q3:',q3)
print("IQR:",IQR)
print("lower_limit:",lower_limit)
print("upper_limit:",upper_limit)
outliers=df[
    (df["Salary"]<lower_limit) |
    (df["Salary"]>upper_limit)
]
print("\n Salary Outliers:",len(outliers))
print("\nSalary outliers:")
print(outliers[["Worker_ID","Salary"]].head(20))
```
