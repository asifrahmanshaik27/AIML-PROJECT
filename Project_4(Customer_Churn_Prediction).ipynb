```python
q1 = df['MonthlyCharges'].quantile(0.25)
q3 = df['MonthlyCharges'].quantile(0.75)

IQR = q3 - q1

lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR
print("Q1: ",q1)
print("Q3: ",q3)
print("IQR: ",IQR)
print("Lower: ",lower_limit)
print("Upper: ",upper_limit)
outliers = df[
    (df["MonthlyCharges"]<lower_limit) |
    (df["MonthlyCharges"]>upper_limit)
]
print("\nNumber of MonthlyCharges outliers: ",len(outliers))
print("\nMonthlyCharges outliers:")
print(outliers[["MonthlyCharges","TotalCharges"]].head(20))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,5))
plt.boxplot(df["MonthlyCharges"])
plt.title=("MonthlyCharges outlier treatment")
plt.ylabel=('MonthlyCharges')
plt.show()
```

```python
average = df.groupby("InternetService")["MonthlyCharges"].mean()
plt.figure(figsize=(8,4))
plt.bar(average.index,average.values,color=["red","yellow"])
plt.title=("Average of MonthlyCharges by InternetService")
plt.xlabel = ("InternetService")
plt.ylabel = ("MonthlyCharges")
plt.show()
```

```python
PaymentMethod_vs_MonthlyCharges = df.groupby("PaymentMethod")["MonthlyCharges"].mean()
print("\nAverage of MonthlyCharges by PaymentMethod:")
print(PaymentMethod_vs_MonthlyCharges)
plt.figure(figsize=(8,4))
plt.plot(
    PaymentMethod_vs_MonthlyCharges.index,
    PaymentMethod_vs_MonthlyCharges.values,
    marker="*"
)

plt.title=("Average of MonthlyCharges by PaymentMethod")
plt.xlabel = ("MonthlyCharges")
plt.ylabel = ("PaymentMethod")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(
    df["tenure"],
    df["MonthlyCharges"],
    alpha=0.6
)

plt.xlabel=("Tenure (Months)")
plt.ylabel=("Monthly Charges")
plt.title=("Tenure vs Monthly Charges")

plt.show()
```

```python
contract = df["Contract"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(contract,
        labels=contract.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title=("contract Distribution")
plt.show()
```
